# VLA integration: DHB-Token

DHB invariants can be tokenized with a causal VQ-VAE for streaming action representation in Vision-Language-Action models.

```python
import numpy as np
import torch
from dhb_xr.encoder.dhb_dr import encode_dhb_dr
from dhb_xr.core.types import DHBMethod
from dhb_xr.tokenization.vqvae import DHBTokenizer

n = 50
np.random.seed(0)
positions = np.cumsum(np.random.randn(n, 3) * 0.01, axis=0)
quaternions = np.tile(np.array([1.0, 0.0, 0.0, 0.0]), (n, 1))
out = encode_dhb_dr(positions, quaternions, method="pos", use_default_initial_frames=True, dhb_method=DHBMethod.DOUBLE_REFLECTION)
invariants = np.concatenate([out["linear_motion_invariants"], out["angular_motion_invariants"]], axis=1).astype(np.float32)
x = torch.from_numpy(invariants).unsqueeze(0)

model = DHBTokenizer(invariant_dim=8, latent_dim=16, codebook_size=64)
indices, reconstructed, z, z_q = model(x)
print("Invariants shape:", x.shape)
print("Token indices shape:", indices.shape)
print("Reconstructed shape:", reconstructed.shape)
print("Sample tokens (first 5):", indices[0, :5].tolist())
```
